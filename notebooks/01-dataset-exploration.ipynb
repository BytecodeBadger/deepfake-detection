```python
path = "../data"
files = []
for root, dirs, file_list in os.walk(path):
    for file in file_list:
        files.append(os.path.join(root, file))

print("Total number of files:", len(files))
print("Sample files:", files[:5])
```

```python
from PIL import Image

# Load Real and Fake images from Train folder
real_path = "../data/Train/Real"
fake_path = "../data/Train/Fake"

real_files = [os.path.join(real_path, f) for f in os.listdir(real_path) if f.endswith('.jpg')][:2]
fake_files = [os.path.join(fake_path, f) for f in os.listdir(fake_path) if f.endswith('.jpg')][:2]

fig, axes = plt.subplots(2, 2, figsize=(10, 10))

# Top row: Real images
for i, ax in enumerate(axes[0]):
    img = Image.open(real_files[i])
    ax.imshow(img)
    ax.axis('off')
    ax.set_title(f"Real: {os.path.basename(real_files[i])}", color='green', fontweight='bold')

# Bottom row: Fake images
for i, ax in enumerate(axes[1]):
    img = Image.open(fake_files[i])
    ax.imshow(img)
    ax.axis('off')
    ax.set_title(f"Fake: {os.path.basename(fake_files[i])}", color='red', fontweight='bold')

plt.tight_layout()
plt.show()
```
